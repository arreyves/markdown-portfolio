```python
The Business Problem: 
DÔVV has a central warehouse and needs to supply 3 branches (Mokolo, Essos, Mendong). 
Each branch has a specific demand for a product, and the transport cost to each branch varies due to traffic and distance in Yaoundé.
The Solution: A linear programming algorithm that calculates the exact amount of stock to send to each branch to minimize transport costs while ensuring shelves are perfectly stocked.
The Tech Stack: pulp (Operations Research library in Python).
The Math:We are solving the transportation problem, minimizing the objective function:$$Z = \sum_{i=1}^{m} \sum_{j=1}^{n} c_{ij} x_{ij}$$Where $c_{ij}$ is the transport cost from warehouse $i$ to 
branch $j$, and $x_{ij}$ is the volume of goods shipped.
```

```python
import pulp  #pulp version

# 1. Setup the Problem: Minimize Transportation Costs
prob = pulp.LpProblem("DOVV_Logistics_Optimization", pulp.LpMinimize)

# 2. Define the Network (Yaoundé Branches)
branches = ["Mokolo", "Essos", "Mendong"]
demand = {"Mokolo": 500, "Essos": 300, "Mendong": 400} # Cartons needed
warehouse_supply = 1500 # Total cartons in Bastos main warehouse

# Transport cost per carton (in FCFA) based on distance/traffic
costs = {"Mokolo": 150, "Essos": 250, "Mendong": 300}

# 3. Decision Variables (How many cartons to send to each branch)
# Minimum 0, Integer values only
route_vars = pulp.LpVariable.dicts("Route", branches, lowBound=0, cat='Integer')

# 4. Objective Function: Minimize Total Cost
prob += pulp.lpSum([route_vars[b] * costs[b] for b in branches]), "Total_Transport_Cost"

# 5. Constraints
# A. We cannot ship more than the warehouse holds
prob += pulp.lpSum([route_vars[b] for b in branches]) <= warehouse_supply, "Supply_Constraint"

# B. Each branch must meet its exact demand
for b in branches:
    prob += route_vars[b] == demand[b], f"Demand_Constraint_{b}"

# 6. Solve the OR Engine
prob.solve()

print(f"Status of Optimization: {pulp.LpStatus[prob.status]}")
print("-" * 30)
for v in prob.variables():
    print(f"Ship to {v.name.replace('Route_', '')}: {v.varValue} cartons")

total_cost = pulp.value(prob.objective)
print("-" * 30)
print(f"Optimized Total Logistics Cost: {total_cost} FCFA")
print("Business Impact: Ensures 100% stock availability at the absolute lowest transport cost.")
```

```python
import numpy as np
from scipy.optimize import linprog #scipy version

# 1. Define the Problem: Transport Costs (Mokolo, Essos, Mendong)
costs = [150, 250, 300]

# 2. Demand Constraints (Equality: Must ship exactly this amount)
# Matrix representing [Mokolo, Essos, Mendong]
A_eq = [
    [1, 0, 0],  # Mokolo needs 500
    [0, 1, 0],  # Essos needs 300
    [0, 0, 1]   # Mendong needs 400
]
b_eq = [500, 300, 400]

# 3. Supply Constraints (Inequality: Cannot ship more than warehouse holds)
A_ub = [[1, 1, 1]] # Sum of all shipments
b_ub = [1500]      # Total cartons in Bastos warehouse

# 4. Bounds (Cannot ship negative cartons)
bounds = [(0, None), (0, None), (0, None)]

# 5. Solve the Operations Research Engine
result = linprog(c=costs, A_ub=A_ub, b_ub=b_ub, A_eq=A_eq, b_eq=b_eq, bounds=bounds, method='highs')

# 6. Output the Business Results
print(f"Status of Optimization: {'Optimal Solution Found' if result.success else 'Failed'}")
print("-" * 30)

branches = ["Mokolo", "Essos", "Mendong"]
for i in range(len(branches)):
    print(f"Ship to {branches[i]}: {int(result.x[i])} cartons")

print("-" * 30)
print(f"Optimized Total Logistics Cost: {int(result.fun)} FCFA")
print("Business Impact: Ensures 100% stock availability at the absolute lowest transport cost.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Data (Sorted by volume)
branches = ["Essos", "Mendong", "Mokolo"]
cartons = [300, 400, 500]
costs = [250, 300, 150] # corresponding route costs in FCFA
total_cost = 270000

# Set aesthetics
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

# Create bar chart
bars = sns.barplot(x=branches, y=cartons, palette="viridis", ax=ax)

# Add Labels and Title
ax.set_title("Operations Research: Optimal Logistics Allocation", fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel("Supermarket Branches", fontsize=12, fontweight='bold')
ax.set_ylabel("Optimal Cartons Allocated", fontsize=12, fontweight='bold')

# Annotate each bar
for i, bar in enumerate(ax.patches):
    ax.annotate(f"{int(bar.get_height())} cartons\n(Route Cost: {costs[i]} FCFA/unit)",
                  (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                  ha='center', va='bottom',
                  size=11, xytext=(0, 5),
                  textcoords='offset points')

# Add a text box summarizing the business impact
props = dict(boxstyle='round', facecolor='#e1f5fe', alpha=0.8)
ax.text(0.05, 0.95, 
        f"Total Minimized Cost: {total_cost:,} FCFA\nBranch Demand Met: 100%\nWarehouse Capacity: Optimized", 
        transform=ax.transAxes, fontsize=11,
        verticalalignment='top', bbox=props)

# Save the plot
plt.tight_layout()
plt.savefig("linkedin_logistics_optimization.png", dpi=300)
```

```python
import matplotlib.pyplot as plt

# Data (Sorted by volume)
branches = ["Essos", "Mendong", "Mokolo"]
cartons = [300, 400, 500]
costs = [250, 300, 150] # corresponding route costs in FCFA
total_cost = 270000

# Setup the figure
fig, ax = plt.subplots(figsize=(10, 6))

# Custom colors to look professional (Teal/Blue palette)
colors = ['#2a9d8f', '#264653', '#e9c46a']

# Create the bar chart using only matplotlib
bars = ax.bar(branches, cartons, color=colors)

# Add Labels and Title
ax.set_title("Operations Research: Optimal Logistics Allocation", fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel("Supermarket Branches", fontsize=12, fontweight='bold')
ax.set_ylabel("Optimal Cartons Allocated", fontsize=12, fontweight='bold')

# Add a subtle grid for readability
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Annotate each bar with the volume and route cost
for i, bar in enumerate(bars):
    ax.annotate(f"{int(bar.get_height())} cartons\n(Route Cost: {costs[i]} FCFA/unit)",
                  (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                  ha='center', va='bottom',
                  size=11, xytext=(0, 5),
                  textcoords='offset points')

# Add the Business Impact summary box
props = dict(boxstyle='round', facecolor='#e1f5fe', alpha=0.8, edgecolor='lightgrey')
ax.text(0.05, 0.95, 
        f"Total Minimized Cost: {total_cost:,} FCFA\nBranch Demand Met: 100%\nWarehouse Capacity: Optimized", 
        transform=ax.transAxes, fontsize=11,
        verticalalignment='top', bbox=props)

# Save the professional plot
plt.tight_layout()
plt.savefig("linkedin_logistics_optimization.png", dpi=300)
print("Plot successfully saved as 'linkedin_logistics_optimization.png'!")
```
